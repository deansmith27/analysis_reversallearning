#  Below is Python code that runs an Integrate and Fire Model, Take some time to investigate the code and then complete the worksheet on Canvas titled "Worksheet 2, Integrate & Fire"

```python
# Jupyter: enable plot rendering inside the notebook (this allows us to actually show the output of the code visually)
%matplotlib inline

import numpy as np
import matplotlib.pyplot as plt

# --- Neuron Parameters (realistic, in mV and pA) ---
C = 200.0 # capacitance
V_th = -55 # threshold voltage, typically where the threshold for AP is
V_reset = -65.0 # Voltage where the system resets because AP is reached
V_rest = -65.0 # voltage where the system is at rest
I_ext = 300.0 #current external
dt = 0.1 #time step


T = 100          # total simulation time in ms
dt = 0.1         # time step in ms
time = np.arange(0, T, dt)  # time vector

# --- Initialize voltage and spike times ---
V = np.full_like(time, V_rest)  # set all values to resting voltage
spike_times = []

# --- Simulation---
for i in range(1, len(time)):
    dV = (I_ext / C) * dt       # ΔV = (I / C) * dt, in mV
    V[i] = V[i-1] + dV

    if V[i] >= V_th:
        V[i-1] = 40.0           # mark spike as +40 mV (for visualization)
        V[i] = V_reset
        spike_times.append(time[i])

# --- Plotting ---
plt.figure(figsize=(10, 4))
plt.plot(time, V, label='fix me')
plt.axhline(V_th, color='r', linestyle='--', label='fix me')
plt.xlabel('time') #what should the x be?
plt.ylabel('Vm(t)') #what should the y be?
plt.title('Intergrate and Fire Model') #what is a good title?
plt.grid(True)
plt.show()

# ---- Raster Plot ---

```

```python
# Deandra Raster Plot with custom data
plt.eventplot(spike_times, colors='black')

plt.yticks([1], [''])

# Label the plot
plt.xlabel('Time (s)') # Label for x
plt.ylabel('Neuron') # Label for y
plt.title('I&F Raster Plot') # Title

plt.show() # Important line! The graph won't be displayed until this is run
```
